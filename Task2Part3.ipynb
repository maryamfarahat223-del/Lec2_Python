```python
from countries import countriesWWWWW
from countries_details_dat import countries_data
```

```python
#1
countries_with_land = []

for country in countriesWWWWW:
    if "land" in country.lower():
        countries_with_land.append(country)

print("Countries containing 'land':")
print(countries_with_land)
```

```python
#2
language_count = {}

for country in countries_data:
    for language in country["languages"]:
        if language in language_count:
            language_count[language] += 1
        else:
            language_count[language] = 1

sorted_languages = sorted(
    language_count.items(),
    key=lambda x: x[1],
    reverse=True
)

print("\n10 most spoken languages:")
for language, count in sorted_languages[:10]:
    print(language, count)
```

```python
#3
sorted_countries = sorted(
    countries_data,
    key=lambda country: country["population"],
    reverse=True
)

print("\n10 most populated countries:")

for country in sorted_countries[:10]:
    print(country["name"], country["population"])
```
