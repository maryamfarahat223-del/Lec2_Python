```python
list_dict =   [
{
    "name": "Ali",
    "Age": 34,
    "Job": "Engineer",
    "Skills": ["python","R","C++"]
}
, 
{
    "name": "Omer",
    "Age": 44,
    "Job": "Tech Engineer",
    "Skills": ["python","C++","Jave"]
}
,
{
    "name": "Ali",
    "Age": 45,
    "Job": "Engineer",
    "Skills": ["python","C"]
}
]
```

```python
my_set =  set()
```

```python
for dict in list_dict:
    for key,value in dict.items():
        if key == "Skills":
            for Skill in value:
                 my_set.add(Skill)
            
print(my_set)
```

```python
my_set =  set()
my_list = []
for dict in list_dict:
    for key,value in dict.items():
        if key == "Skills":
            my_list.extend(value)
        
      
my_set = set(my_list)  
print(my_set)
```

```python
my_set =  set()
for dict in list_dict:
    for key,value in dict.items():
        if key == "Skills":
            my_set.update(value)
        
print(my_set)
```
